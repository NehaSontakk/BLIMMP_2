```python
!pip install bioservices

import time
from bioservices import KEGG
import pandas as pd

kegg = KEGG()

def fetch_ko_to_reaction_mapping(module_id):
    """Fetch KO → reaction mapping for a KEGG module ID."""
    try:
        module_info = kegg.get(module_id)
        if not module_info:
            return {}

        module_parsed = kegg.parse(module_info)
        ko_to_reaction = {}

        if "ORTHOLOGY" in module_parsed:
            for ko_group, description in module_parsed["ORTHOLOGY"].items():
                if "[RN:" in description:
                    # Extract reaction section
                    reaction_section = description.split("[RN:")[-1]
                    reaction_numbers = reaction_section.replace("]", "").split()

                    # Assign each KO in the group
                    for ko in ko_group.split(","):
                        ko = ko.strip()
                        if not ko.startswith("K"):
                            continue
                        ko_to_reaction.setdefault(ko, []).extend(reaction_numbers)

        return ko_to_reaction

    except Exception as e:
        print(f"Error fetching {module_id}: {e}")
        return {}

# ---- Apply to your Excel file ----
data = pd.read_excel(
    "/content/drive/MyDrive/Lab Work/Bayesian_Graph_Metabolic_Paper/KEGG_Graph_Generation/KEGG_Prokaryote_Modules_with_Definitions.xlsx"
)

all_mappings = {}

for module_id in data["Module ID"].dropna().unique():
    all_mappings[module_id] = fetch_ko_to_reaction_mapping(module_id)
    time.sleep(0.2)  # be gentle with KEGG API

# Convert nested dict into a dataframe
rows = []
for module_id, mapping in all_mappings.items():
    for ko, reactions in mapping.items():
        rows.append({
            "Module ID": module_id,
            "KO": ko,
            "Reactions": ";".join(reactions)
        })

mapping_df = pd.DataFrame(rows)
mapping_df.to_excel("/content/drive/MyDrive/Lab Work/Bayesian_Graph_Metabolic_Paper/KEGG_Neighbors_Generation/KO_to_Reactions.xlsx", index=False)

print("Saved KO→reaction mapping to KO_to_Reactions.xlsx")
```
